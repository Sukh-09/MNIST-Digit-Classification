```python
def ink_density(image):
    normalized = image / 255.0
    return np.mean(normalized)

# Test on the first filtered image
density = ink_density(X_filtered[0])

print("Digit:", y_filtered[0])
print("Ink density:", density)
```

```python
def left_right_balance(image):
    image = image.reshape(28, 28) / 255.0

    left = np.sum(image[:, :14])
    right = np.sum(image[:, 14:])

    balance = (left - right) / (left + right + 1e-8)

    return balance
```

```python
def top_bottom_balance(image):
    image = image.reshape(28, 28) / 255.0

    top = np.sum(image[:14, :])
    bottom = np.sum(image[14:, :])

    balance = (top - bottom) / (top + bottom + 1e-8)

    return balance
```

```python
def extract_features(image):
    f1 = ink_density(image)
    f2 = left_right_balance(image)
    f3 = top_bottom_balance(image)

    return np.array([f1, f2, f3])

X_features = np.array([
    extract_features(image)
    for image in X_filtered
])

print("Feature matrix shape:", X_features.shape)
print("First feature vector:", X_features[0])
```

```python
# Create binary labels for each classifier
y_0 = (y_filtered == 0).astype(int)
y_5 = (y_filtered == 5).astype(int)
y_8 = (y_filtered == 8).astype(int)

print("0-detector:", np.unique(y_0))
print("5-detector:", np.unique(y_5))
print("8-detector:", np.unique(y_8))
```

```python
class Perceptron:
    def __init__(self, learning_rate=0.01):
        self.lr = learning_rate
        self.weights = None
        self.bias = 0.0

    def predict(self, X):
        z = X @ self.weights + self.bias
        return (z >= 0).astype(int)

    def fit(self, X, y, epochs=1000):
        n_samples, n_features = X.shape

        self.weights = np.zeros(n_features)
        self.bias = 0.0

        self.training_errors = []

        for epoch in range(epochs):
            mistakes = 0

            for i in range(n_samples):
                z = np.dot(X[i], self.weights) + self.bias
                prediction = int(z >= 0)

                error = y[i] - prediction

                if error != 0:
                    self.weights += self.lr * error * X[i]
                    self.bias += self.lr * error

                mistakes += int(error != 0)

            self.training_errors.append(
                mistakes / n_samples
            )

        return self
```

```python
from sklearn.model_selection import StratifiedKFold

# Standard MNIST training portion: first 60,000 images
X_train = X[:60000]
y_train = Y[:60000]

# Keep only digits 0, 5, and 8
train_mask = np.isin(y_train, [0, 5, 8])

X_train_filtered = X_train[train_mask]
y_train_filtered = y_train[train_mask]

# Extract our three features
X_cv = np.array([
    extract_features(image)
    for image in X_train_filtered
])

# Create five folds
skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

folds = list(skf.split(X_cv, y_train_filtered))

print("Feature matrix:", X_cv.shape)
print("Number of folds:", len(folds))
```

```python
from numba import njit

@njit
def train_perceptron_cv(X_train, y_train, X_val, y_val,
                        epochs=1000, learning_rate=0.01):

    weights = np.zeros(X_train.shape[1])
    bias = 0.0

    train_errors = np.zeros(epochs)
    val_errors = np.zeros(epochs)

    for epoch in range(epochs):

        # Train for one epoch
        for i in range(len(X_train)):
            z = np.dot(X_train[i], weights) + bias
            prediction = 1 if z >= 0 else 0
            error = y_train[i] - prediction

            if error != 0:
                weights += learning_rate * error * X_train[i]
                bias += learning_rate * error

        # Evaluate after the epoch
        train_mistakes = 0
        val_mistakes = 0

        for i in range(len(X_train)):
            prediction = 1 if np.dot(X_train[i], weights) + bias >= 0 else 0
            if prediction != y_train[i]:
                train_mistakes += 1

        for i in range(len(X_val)):
            prediction = 1 if np.dot(X_val[i], weights) + bias >= 0 else 0
            if prediction != y_val[i]:
                val_mistakes += 1

        train_errors[epoch] = train_mistakes / len(X_train)
        val_errors[epoch] = val_mistakes / len(X_val)

    return train_errors, val_errors
```

```python
results = {}

for digit in [0, 5, 8]:

    print(f"Training {digit}-detector...")

    # Binary labels for the selected digit
    y_binary = (y_train_filtered == digit).astype(np.int64)

    all_train_errors = []
    all_val_errors = []

    for fold_number, (train_idx, val_idx) in enumerate(folds):

        X_fold_train = X_cv[train_idx]
        X_fold_val = X_cv[val_idx]

        y_fold_train = y_binary[train_idx]
        y_fold_val = y_binary[val_idx]

        train_err, val_err = train_perceptron_cv(
            X_fold_train,
            y_fold_train,
            X_fold_val,
            y_fold_val,
            epochs=1000,
            learning_rate=0.01
        )

        all_train_errors.append(train_err)
        all_val_errors.append(val_err)

        print(f"  Fold {fold_number + 1}/5 completed")

    results[digit] = {
        "train": np.mean(all_train_errors, axis=0),
        "val": np.mean(all_val_errors, axis=0)
    }

print("All perceptron training completed!")
```

```python
fig, axes = plt.subplots(1, 3, figsize=(17, 5))

for ax, digit in zip(axes, [0, 5, 8]):

    ax.plot(
        range(1, 1001),
        results[digit]["train"],
        label="Average Training Error"
    )

    ax.plot(
        range(1, 1001),
        results[digit]["val"],
        label="Average Validation Error"
    )

    ax.set_title(f"Perceptron: {digit}-Detector")
    ax.set_xlabel("Epoch")
    ax.set_ylabel("Classification Error")
    ax.legend()
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig("perceptron_learning_curves.png", dpi=300)
plt.show()
```

```python
def train_linear_regression(X_train, y_train, X_val, y_val,
                            epochs=1000, learning_rate=0.1):

    # Add a column of ones for the bias
    X_train_b = np.column_stack([
        np.ones(len(X_train)), X_train
    ])

    X_val_b = np.column_stack([
        np.ones(len(X_val)), X_val
    ])

    # Initialize weights and bias to zero
    weights = np.zeros(X_train_b.shape[1])

    train_errors = np.zeros(epochs)
    val_errors = np.zeros(epochs)

    n = len(X_train)

    for epoch in range(epochs):

        # Continuous predictions
        predictions = X_train_b @ weights

        # Calculate gradient of mean squared error
        gradient = (2 / n) * (
            X_train_b.T @ (predictions - y_train)
        )

        # Gradient descent update
        weights -= learning_rate * gradient

        # Convert continuous outputs to binary predictions
        train_pred = (X_train_b @ weights >= 0.5).astype(int)
        val_pred = (X_val_b @ weights >= 0.5).astype(int)

        # Calculate classification errors
        train_errors[epoch] = np.mean(train_pred != y_train)
        val_errors[epoch] = np.mean(val_pred != y_val)

    return train_errors, val_errors
```

```python
lr_results = {}

for digit in [0, 5, 8]:

    print(f"Training linear regression {digit}-detector...")

    y_binary = (y_train_filtered == digit).astype(int)

    all_train_errors = []
    all_val_errors = []

    for fold_number, (train_idx, val_idx) in enumerate(folds):

        X_fold_train = X_cv[train_idx]
        X_fold_val = X_cv[val_idx]

        y_fold_train = y_binary[train_idx]
        y_fold_val = y_binary[val_idx]

        train_err, val_err = train_linear_regression(
            X_fold_train,
            y_fold_train,
            X_fold_val,
            y_fold_val,
            epochs=1000,
            learning_rate=0.1
        )

        all_train_errors.append(train_err)
        all_val_errors.append(val_err)

        print(f"  Fold {fold_number + 1}/5 completed")

    lr_results[digit] = {
        "train": np.mean(all_train_errors, axis=0),
        "val": np.mean(all_val_errors, axis=0)
    }

print("Linear regression training completed!")
```

```python
fig, axes = plt.subplots(1, 3, figsize=(17, 5))

for ax, digit in zip(axes, [0, 5, 8]):

    ax.plot(
        range(1, 1001),
        lr_results[digit]["train"],
        label="Average Training Error"
    )

    ax.plot(
        range(1, 1001),
        lr_results[digit]["val"],
        label="Average Validation Error"
    )

    ax.set_title(f"Linear Regression: {digit}-Detector")
    ax.set_xlabel("Epoch")
    ax.set_ylabel("Classification Error")
    ax.legend()
    ax.grid(True, alpha=0.3)

plt.tight_layout()

plt.savefig(
    "linear_regression_learning_curves.png",
    dpi=300
)

plt.show()
```
